```python
import pandas as pd
glucose = pd.read_csv("../../raw_data/Glucose_measurements.csv")

print(glucose.shape)
print(glucose.head())
print(glucose.info())
print(glucose.isnull().sum())
```

```python
glucose['Patient_ID'].nunique()
```

```python
glucose['Measurement'].describe()
```

```python
patient_id = glucose['Patient_ID'].iloc[0]

patient = glucose[glucose['Patient_ID'] == patient_id].copy()

patient.head(20)
```

```python
len(patient)
```

```python
patient['timestamp'] = pd.to_datetime(
    patient['Measurement_date'] + ' ' + patient['Measurement_time']
)
```

```python
patient[['timestamp','Measurement']].head(20)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,5))
plt.plot(patient['timestamp'],patient['Measurement'])
plt.xlabel('Time')
plt.ylabel('Glucose measurement')
plt.title(f'Glucose trajectory — {patient_id}')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
patient['time_diff'] = patient['timestamp'].diff().dt.total_seconds() / 60

patient['time_diff'].value_counts().head(10)
```

```python
patient['time_diff'].describe()
```

```python
measurements_per_patient = glucose.groupby('Patient_ID').size()

measurements_per_patient.describe()
```

```python
measurements_per_patient.sort_values().head(10)
```

```python
measurements_per_patient.sort_values(ascending=False).head(10)
```

```python
patient[patient['time_diff'] == 0][
    ['timestamp','Measurement']
].head(20)
```

```python
patient['time_diff'].value_counts().sort_index().tail(20)
```

```python
(patient['time_diff'] > 15).sum()
```

```python
patient['hour'] = patient['timestamp'].dt.hour
```

```python
patient.groupby('hour')['Measurement'].mean()
```

```python
one_week = patient[
    patient['timestamp'] <
    patient['timestamp'].min() + pd.Timedelta(days=7)
]

plt.figure(figsize=(14,5))
plt.plot(one_week['timestamp'],one_week['Measurement'])
plt.xlabel('Time')
plt.ylabel('Glucose')
plt.title(f'One-week glucose trajectory — {patient_id}')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sample_ids = glucose['Patient_ID'].drop_duplicates().sample(3,random_state=42)

for pid in sample_ids:
    p = glucose[glucose['Patient_ID'] == pid].copy()
    p['timestamp'] = pd.to_datetime(
        p['Measurement_date'] + ' ' + p['Measurement_time']
    )
    
    start = p['timestamp'].min()
    p = p[p['timestamp'] < start + pd.Timedelta(days=3)]
    
    plt.figure(figsize=(14,4))
    plt.plot(p['timestamp'],p['Measurement'])
    plt.xlabel('Time')
    plt.ylabel('Glucose')
    plt.title(f'3-day glucose trajectory — {pid}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

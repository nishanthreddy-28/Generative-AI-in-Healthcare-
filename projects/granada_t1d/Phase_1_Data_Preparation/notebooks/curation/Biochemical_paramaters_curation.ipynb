```python
import pandas as pd

path = 'Originals/'
path_to_save = 'Data curated/'

file = 'A_LibreviewResultadosAnaliticos.txt'
df_original = pd.read_csv(path + file, sep='|', encoding='latin-1' )
```

# Drop variables

## Irrelevant variable

```python
# Prueba_Descripción_Corta
df_original = df_original.drop(labels=['Prueba_Descripción_Corta'], axis=1)
```

## Empty variables

```python
# Resultado_Descripción and Texto_Resultado
df_original = df_original.drop(labels=['Resultado_Descripción','Texto_Resultado'], axis=1)
```

# Drop records

## Without results
A hyphen (-) appear as no result

```python
with_hyphen_filter = (df_original['Resultado'] == '-')
df_original = df_original[~with_hyphen_filter]
df_original.reset_index(inplace=True, drop=True)
```

## Duplicated

```python
dup = df_original.duplicated()
df_original = df_original[~dup]
df_original.reset_index(inplace=True, drop=True)
```

# Translation of values

## Columns

```python
column_name_english = ['Patient_ID', 'Reception_date','Name', 'Value']
df_original.columns = column_name_english
```

## Values of 'Name' column

```python
dict_es_to_en = {'Name' :
                       {'Creatinina' : 'Creatinine',
                        'Glucosa' : 'Glucose (mg/dl)',
                        'Potasio' : 'Potassium',
                        'Sodio' : 'Sodium',
                        'Alanina transaminasa (GPT)' : 'Alanine transaminase (GPT)',
                        'Gamma glutamiltransferasa (GGT)' : 'Gamma-glutamyl Transferase (GGT)',
                        'Colesterol total' : 'Total cholesterol',
                        'Triglicéridos' : 'Triglycerides',
                        'Colesterol de HDL' : 'HDL cholesterol',
                        'Hemoglobina glicosilada (A1c)' : 'Glycated hemoglobin (A1c)',
                        'Acido úrico' : 'Uric acid',
                        'Creatinina (orina)' : 'Creatinine (urine)',
                        'Albúmina (orina)' : 'Albumin (urine)',
                        'Tirotropina (TSH)' : 'Thyrotropin (TSH)',
                        'Cloro' : 'Chlorine',
                        'Ac anti IA2' : 'IA2 ANTIBODIES',
                        'Insulina' : 'Insulin' }}

df_original.replace(dict_es_to_en, inplace=True)
```

# Convert “Result” into float type

```python
df_original['Value'] = df_original['Value'].str.replace(',','.')
df_original['Value'] = df_original['Value'].astype(float)
```

# Convert Reception_date column into datetime format

```python
df_original['Reception_date'] = pd.to_datetime(df_original['Reception_date'], dayfirst=True)
```

# Sort the dataset by Patient_ID and Reception_date

```python
df_original = df_original.sort_values(by=['Patient_ID', 'Reception_date'])
df_original.reset_index(inplace=True, drop=True)
```

# Save data as CSV

```python
df_original.to_csv(path_to_save + 'Biochemical_parameters.csv',index=False)
```

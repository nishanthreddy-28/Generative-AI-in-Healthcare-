```python
import pandas as pd
import numpy as np

path = 'Originals/'
path_to_save = 'Data curated/'

file = 'A_LibreviewMedidasGlucosa.txt'
df_original = pd.read_csv(path + file, sep='|')
```

# Eliminate entries without glucose measures

```python
df_original.dropna(subset=['Medida'], axis=0, inplace=True) # Drop the rows where are NaN values in the column "Medida"
```

# Convert glucose measure to short type
Convert the type of the glucose measurement column in short to better space optimization


```python
df_original['Medida'] = df_original['Medida'].astype(dtype=np.short)
```

# Translate the column names

```python
column_name_english = ['ID', 'Measurement_date', 'Measurement']
df_original.columns = column_name_english
```

# Drop duplicated rows

```python
dup = df_original.duplicated()
df_original = df_original[~dup]
df_original.reset_index(inplace=True, drop=True)
```

# Convert column "Measurement_date" to Timestamp format

```python
df_original['Measurement_date'] = pd.to_datetime(df_original['Measurement_date'], dayfirst=True)
```

# Split dates in two columns

```python
# Create a column Time calculating time form the Measurement_date column.
df_original['Measurement_time'] = df_original['Measurement_date'].apply(func=lambda x: x.time())

# Create a column Date calculating date form the Measurement_date column.
df_original['Measurement_time'] = df_original['Measurement_time'].apply(func=lambda x: x.date())

# Putting a better position to the columns
df_original = df_original[['Patient_ID', 'Measurement_date', 'Measurement_time', 'Measurement']]
```

# Sort by 'Patient_ID', 'Measurement_date', and 'Measurement_time'

```python
df_original = df_original.sort_values(by=['Patient_ID', 'Measurement_date', 'Measurement_time'])
df_original.reset_index(inplace=True, drop=True)
```

# Save data as CSV

```python
df_original.to_csv(path_to_save + 'Glucose_measurements.csv',index=False)
```

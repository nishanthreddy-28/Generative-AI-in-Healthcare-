```python
import pandas as pd

path = 'Originals/'
file = 'DatosPacientes.txt'

df_original = pd.read_csv(path + file, sep='|', encoding='latin-1' )
```

# Drop column UF
(Irrelevant, all rows have the same values)

```python
df_relevant = df_original.drop(labels=['UF'], axis=1)
df_relevant
```

# Translation and transformation
## Column names translation

```python
df_relevant_translated = df_relevant.copy()
column_name_english = ['Patient_ID', 'Sex', 'Birth_year']
df_relevant_translated.columns = column_name_english
df_relevant_translated
```

## Column order

```python
df_relevant_translated = df_relevant_translated[['Patient_ID', 'Sex', 'Birth_year']]
```

## Values of Variable 'sex' transformation
BEFORE
M-> Mujeres and H-> Hombres
NOW
M-> F (Female) and H-> M (Male)

```python
df_relevant_translated.replace({'Sex': {'M' : 'F', 'H' : 'M'}}, inplace=True)
```

# Drop duplicated rows

```python
dup = df_relevant_translated.duplicated()
df_relevant_translated = df_relevant_translated[~dup]
df_relevant_translated.reset_index(inplace=True, drop=True)
```

# Sort dataset by ID

```python
df_relevant_translated = df_relevant_translated.sort_values(by=['Patient_ID'])
df_relevant_translated.reset_index(inplace=True, drop=True)
```

# Save data as CSV

```python
df_relevant_translated.to_csv('Patient_info_original.csv', index=False)
```
